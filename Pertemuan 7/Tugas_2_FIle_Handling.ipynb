```python
def createFile(namaFile):
    file = open(namaFile, "w")
    teks = input("Masukkan teks kamu: ")
    file.write(teks)
    file.close()
    print("File berhasil dibuat! / ditulis\n")


def readFile(namaFile):
    file = open(namaFile, "r")
    isi = file.read()
    print("\n===== Isi File =====")
    print(isi)
    file.close()
    print("")


def appendFile(namaFile):
    file = open(namaFile, "a")
    teks = input("Masukkan teks tambahan: ")
    file.write("\n" + teks)
    file.close()
    print("Teks berhasil ditambahkan\n")


while(True):
    print("===== File Handling Program =====")
    print("1. Buat dan Tulis File")
    print("2. Baca File")
    print("3. Tambah Teks pada File")
    print("4. Keluar Program")

    option = input("Pilih menu (1/2/3/4): ")

    if(option == "1"):
        namaFile = input("Masukkan nama file: ")
        createFile(namaFile)

    elif(option == "2"):
        namaFile = input("Masukkan nama file: ")
        readFile(namaFile)

    elif(option == "3"):
        namaFile = input("Masukkan nama file: ")
        appendFile(namaFile)

    elif(option == "4"):
        print("Anda telah keluar dari program. Terima kasih.")
        break

    else:
        print("Pilihan tidak valid\n")
```
